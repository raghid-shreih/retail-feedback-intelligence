# Retail Feedback Intelligence walkthrough

This notebook demonstrates local inspection and the package interface. Download the Women’s E-Commerce Clothing Reviews CSV from Kaggle and place it in `data/`. No data or API responses are embedded here.

````python
from pathlib import Path
from retail_feedback.data import load_reviews, sample_reviews

DATA_PATH = Path("../data/Womens Clothing E-Commerce Reviews.csv")
if DATA_PATH.exists():
    reviews = load_reviews(DATA_PATH)
else:
    reviews = None
    print(f"Download the CSV first: {DATA_PATH}")
````

````python
if reviews is not None:
    print("Reviews with text:", len(reviews))
    print("Columns:", list(reviews.columns))
    print("Recommendation distribution:")
    print(reviews["recommended_ind"].value_counts(dropna=False))
````

## Reproducible sample

The original notebook sampled 50 reviews with seed 42. Sampling depends on the exact source file and retained row order.

````python
if reviews is not None:
    sample = sample_reviews(reviews, 50, seed=42)
    print("Sample size:", len(sample))
    print("Source row indices:", sample.index.tolist()[:10])
````
